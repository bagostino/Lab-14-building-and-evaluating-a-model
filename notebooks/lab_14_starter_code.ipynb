## Part 2.

Simple Linear Regression.

* Regress one numeric feature on the target variable `charges`
* Set up your `X` and `y` first.
* Do a train test split, default test size is fine. MAKE SURE TO SET A RANDOM STATE.
* Fit the model.
* Interpret the coefficient.

````python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

````

````python
# Create X and y
X = df[['bmi']]
y = df['charges']

````

````python
# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

````

````python
# create a new instance of LinearRegression
lr = LinearRegression()

# fit it on your training data.
lr.fit(X_train, y_train)

# make predictions store as y_pred
y_pred = lr.predict(X_test)

````

````python
# Evaluate your model with MAE, RMSE, R2 Score
print('MAE: ', mean_absolute_error(y_test, y_pred))
print('RMSE:', root_mean_squared_error(y_test, y_pred))
print('R2:  ', r2_score(y_test, y_pred))

````

````python
# Interpret the coefficient
print(f'Coefficient: {lr.coef_[0]:.2f}')
print(f'Intercept:   {lr.intercept_:.2f}')
print(f'\nFor every single point increase in BMI, predicted charges increase by ${lr.coef_[0]:.2f}')
````

## Part 2.

Multiple Linear Regression with numeric features.

* Pick at least 3 of the numeric features in the dataset.
* Recreate your feature matrix, `X`
* Do another train test split.
* Make a Multiple Linear Regression.
* Evaluate your model.
* Interpret a coefficient (Remember the difference for interpreting coefficients of multiple linear regression vs. simple!)


````python
# define a list of numeric features
numeric_features = ['bmi', 'age', 'children']

# define X, y
X = df[numeric_features]
y = df['charges']

````

````python
# train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

````

````python
# create a new instance of linear regression
mlr = LinearRegression()

# fit on the training data
mlr.fit(X_train, y_train)

# make predictions
y_pred = mlr.predict(X_test)
````

````python
# Evaluate based on metrics
print('MAE: ', mean_absolute_error(y_test, y_pred))
print('RMSE:', root_mean_squared_error(y_test, y_pred))
print('R2:  ', r2_score(y_test, y_pred))

````

````python
# interpret coefficients
for feature, coef in zip(numeric_features, mlr.coef_):
    print(f'{feature}: {coef:.2f}')
print(f'\nHolding all other features constant:')
print(f'  A 1-unit increase in BMI is associated with a ${mlr.coef_[0]:.2f} change in charges.')
print(f'  A 1-year increase in age is associated with a ${mlr.coef_[1]:.2f} change in charges.')
print(f'  One additional child is associated with a ${mlr.coef_[2]:.2f} change in charges.')

````

### Add Categorical Variables.

* Select numeric and categorical variables for your feature matrix
* Encode the categorical variables using `pd.get_dummies()`
  * REMEMBER: There is some arguments you need to adjust inside of `pd.get_dummies()`
* Do a train test split
* Train and evaluate the model
* BONUS: Interpret a coefficient from one of the encoded categorical variables.

````python
# Define features, X, y
features = ['age', 'bmi', 'children', 'sex', 'smoker', 'region']
X = df[features]
y = df['charges']

````

````python
# Encode your categorical variables.
X = pd.get_dummies(X, drop_first=True)
X.head()

````

````python
# Do a train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# instantiate a new model
model = LinearRegression()

# fit the model
model.fit(X_train, y_train)

# make predictions
y_pred = model.predict(X_test)


 
````

````python
# Evalueate it
print('MAE: ', mean_absolute_error(y_test, y_pred))
print('RMSE:', root_mean_squared_error(y_test, y_pred))
print('R2:  ', r2_score(y_test, y_pred))
````

## Bonus Section: Comparing Training and Test Performance

So far, we have evaluated how well our model performs on the test data.

We can also check how well the model performs on the training data. This helps us compare:

- How well the model performs on data it learned from
- How well the model performs on new, unseen data

In `sklearn`, regression models have a `.score()` method.

For linear regression, `.score()` returns the model's **R² score**.

R² measures how much of the variation in the target variable is explained by the model.

Run the following code:

```python
print(f"Training score: {model.score(X_train, y_train):.3f}")
print(f"Testing score: {model.score(X_test, y_test):.3f}")
```

### How to Interpret the Scores

If the training score is much higher than the testing score, the model may be overfitting.

This means the model performs well on the data it learned from, but does not perform as well on new, unseen data.

If the training score and testing score are both low, the model may be underfitting.

This means the model is not learning enough of the relationship between the features and the target variable.

If the training score and testing score are close together, and both scores are reasonably strong for the problem, the model may be generalizing well.

This means the model is learning useful patterns from the training data and still performing well on new data.
